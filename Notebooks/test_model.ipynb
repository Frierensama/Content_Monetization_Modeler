```python
import pickle
from dotenv import load_dotenv
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
load_dotenv()
```

```python
with open(os.getenv('model_path'), 'rb') as file:
    lr_model = pickle.load(file) 
```

```python
lr_model
```

```python
df = pd.read_csv("C:/Users/rudeu/Documents/Github_Repos/Content_Monetization_Modeler/Data/model_data.csv")
```

```python
X = df.drop(columns=['ad_revenue_usd'])
y = df['ad_revenue_usd']
```

```python
coef_df = pd.DataFrame({'Feature': X.columns, 'Coefficient': lr_model.coef_})
```

```python
coef_df.sort_values(by='Coefficient', key=abs, ascending=False)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

top_features = coef_df.sort_values(by='Coefficient', key=abs, ascending=False).head(10)

plt.figure(figsize=(10,10))

sns.barplot(data=top_features,x='Coefficient',y='Feature')

plt.title('Top Influential Features')
plt.show()
```
